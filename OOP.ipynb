```python
from abc import ABC, abstractmethod
class Employee(ABC):
    def __init__(self,name):
        self.name = name
        self.__salary = 0
       
    def get_salary(self):
        return self.__salary

    def _set_salary(self,salary):
        self.__salary = salary

    @abstractmethod
    def calculate_salary(self,salary):
        pass        

class FullTimeEmployee(Employee):
    def __init__(self,name,monthly_salary,bonus):
        super().__init__(name)
        self.monthly_salary = monthly_salary
        self.bonus = bonus
        
    def calculate_salary(self):
        total = self.monthly_salary + self.bonus
        self._set_salary(total)
        return total

class PartTimeEmployee(Employee):
    def __init__(self,name,hourly_rate,no_hours):
        super().__init__(name)
        self.hourly_rate = hourly_rate
        self.no_hours = no_hours

    def calculate_salary(self):
        total = self.hourly_rate * self.no_hours
        self._set_salary(total)
        return total

emp1 = FullTimeEmployee('Mark',60000,10000)
emp2 = PartTimeEmployee('zeff',6000,360)

emp1.calculate_salary()
emp2.calculate_salary()

print(emp1.name, "Salary:", emp1.get_salary())
print(emp2.name, "Salary:", emp2.get_salary())        
```

```python
from abc import ABC, abstractmethod

class Account(ABC):
    def __init__(self,name,accno,balance):
        self.name = name
        self.accno = accno
        self.__balance = 40000
        
    def get_balance(self):
        return self.__balance

    def _update_balance(self,amt):
        self.__balance += amt    

    @abstractmethod
    def withdraw(self,amt):
        pass

    def deposit(self, amt):
        self._update_balance(amt)
        print("Balance Updated.",amt)

class SavingAccount(Account):
    
    def withdraw(self,amt):
        if amt <= self.get_balance():
            self._update_balance(-amt)
            print("Withdrawal from savings: ",amt)
        else:
            print("Insufficient balance.")
    
    def add_interest(self):
        interest = self.get_balance() * 0.07
        self._update_balance(interest)
        print("Updated balance is: ", interest)

class CurrentAccount(Account):

    def withdraw(self,amt):
        if amt <= self.get_balance() + 4000:
            self._update_balance(-amt)
            print("Withdraw from Current: ",amt)
        else:
            print("Limit exceded.")

s = SavingAccount(101,'Suchit',15000)
s.deposit(2000)
s.withdraw(5000)
s.add_interest()

c = CurrentAccount(102,'Suchitra',5000)
c.withdraw(8000)    
```

```python
class dCart:
    def __init__(self):
        self.items = []
    def additems(self,itemname,unitprice,qty):
        choice = (itemname, unitprice, qty)
        self.items.append(choice)
    def show_bill(self):
        total = 0
        for i in self.items:
            total += (i[1] * i[2])
        return total    
myCart = dCart()
myCart.additems('Laptop', 50000, 2)
myCart.additems('Mobile',15000,5)
myCart.additems('Bike',300000,2)
myCart.show_bill()
print(f"Your bill is {myCart.show_bill()} INR")
```

```python
# 1) Inheritance - Attributes and methods of one class can be used by another class.
# benefits - reusability of code.
# parent class / base class / supper class from which attributes and methods are passed on to.
# child class / derived class - 
# object of parent class is not created usually.
# Types - 1. single - one super class and one derived class.
# 2. multilevel - one parent having one child class which is parent class for another class.
# 3. multiple inheritance - one child class having more than one parent class.
```

```python
class father:
    def __init__(self,name,profession):
        self.fname = name
        self.profession = profession

class son(father):
    def __init__(self, name, profession, sname, age):
        father.__init__(self, name, profession)  #Called to parent class constructor
        self.sname = sname
        self.age = age
        
    def show_details(self):
        print(f"I am {self.sname} son of {self.fname} a {self.profession} by profession")

s1 = son("Amit", "officer","Laksh",6)
s1.show_details()
```

```python
class grandfather:
    def __init__(self, gname):
        self.gname = gname

class father(grandfather):
    def __init__(self, gname, fname):
        grandfather.__init__(self,gname)
        self.fname = fname

class son(father):
    def __init__(self, gname, fname, sname):
        father.__init__(self,gname,fname)
        self.sname = sname
    def display(self):
        print(f'I am {self.sname} son of {self.fname} and grandson of {self.gname}')

s1 = son('Shahaji','Shivaji','Sambhaji')
s1.display()        
```

```python
class father:
    def __init__(self, fname):
        self.fname = fname

class mother:
    def __init__(self,mname):
        self.mname = mname

class son(father,mother):
    def __init__(self,fname,mname,sname):
        father.__init__(self,fname)
        mother.__init__(self,mname)
        self.sname = sname

    def display(self):
        print(f'I am {self.sname} son of {self.mname} and {self.fname}')

s = son('Amit','Pratiksha','Laksh')
s.display()
```

```python
class shape:
    def __init__(self,nos,color):
        self.nos = nos
        self.color = color

class rectangle(shape):
    def __init__(self,nos,color, length, breadth):
        shape.__init__(self,nos, color)
        self.length = length
        self.breadth = breadth

    def area(self):
        print(f'Area of shape {self.length * self.breadth}')

r = rectangle(5,'Red',3.7,11.5)
r.area()
```

```python
import math
class Circle:
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius ** 2
    def circumference(self):
        return 2 * math.pi * self.radius
c = Circle(3)
print(c.area())
print(c.circumference())
```

```python
from abc import ABC, abstractmethod
class Person:
    def __init__(self,name,age):
        self.name = name
        self.__age = age
    def get_age(self):
        return self.__age
    @abstractmethod
    def get_role(self):
        pass

class Doctor(Person):
    def __init__(self,name,age,fee,specialization):
        super().__init__(name,age)
        self.fee = fee
        self.specialization = specialization
    def get_role(self):
        return 'Doctor'
    def treat_patient(self, patient):
        print(f'{self.name} is treating {patient.name}')

class Patient(Person):
    def __init__(self,name,age,disease,bill):
        super().__init__(name,age)
        self.disease = disease
        self.__bill = bill

    def get_role(self):
        return 'Patient'

    def pay_bill(self):
        print(f'{self.name} paid bill: {self.__bill}')

d = Doctor("Dr.Laksh",27,3000,'Orthologist')
p1 = Patient("Manoj Solanki",60,"complex spinal reconstruction, total knee replacement, heel bone (calcaneus) fracture surgery",25000000)
p2 = Patient("Pankaj Solanki",59,"complex spinal reconstruction, total knee replacement, heel bone (calcaneus) fracture surgery, Angioplasty",35000000)

d.treat_patient(p1)
d.treat_patient(p2)
print(p1.get_age(), p1.name)
print(p2.get_age(), p2.name)
p1.pay_bill()
p2.pay_bill()

people = [d,p1,p2]

for person in people:
    print(person.name, "is a", person.get_role())
```

```python
from abc import ABC, abstractmethod
class User:
    def __init__(self,name,phone):
        self.name = name
        self.__phone = phone
    def get_phone(self):
        return self.__phone
    @abstractmethod
    def get_user_type(self):
        pass

class Customer(User):
    def __init__(self,name,phone):
        super().__init__(name,phone)
        self.cart = []
    def get_user_type(self):
        return 'Customer'
    def order_food(self,item):
        self.cart.append(item)
        print(f'{self.name} ordered {item}')

class DeliveryPartner(User):
    def __init__(self, name, phone, vehicle):
        super().__init__(name, phone)
        self.vehicle = vehicle

    def get_user_type(self):
        return "Delivery Partner"

    def deliver_order(self, customer):
        print(f"{self.name} delivered order to {customer.name}")

class Restaurant:
    def __init__(self, name, menu):
        self.name = name
        self.menu = menu

    def show_menu(self):
        print(f"Menu of {self.name}: {self.menu}")

r1 = Restaurant("Pizza Hut", ["Pizza", "Burger"])
c1 = Customer("Vishal", "1234567890")
d1 = DeliveryPartner("Aakash", "9876543210", "Bike")

r1.show_menu()
c1.order_food("Pizza")
d1.deliver_order(c1)

users = [c1, d1]
for user in users:
    print(user.name, "is a", user.get_user_type())
```
