```python
num = [2,4,5,7]
cube = list(map(lambda x: x**3, num))
print(cube)
```

```python
num = [11,23,44,56,7]
odd = list(filter(lambda x: x%2 == 1,num))
print(odd)
```

```python
names = ["Sidhharth","laksh","Amit","priti","Anjali"]
sorted_names = sorted(names, key=lambda x:x.lower())
sorted_name = sorted(names, key=lambda x:len(x.lower()), reverse = True)
print(sorted_names)
print(sorted_name)
```

```python
pairs = [(3,4,9),(6,7,6),(2,8,23),(31,2,11)]
sorted_pairs = sorted(pairs, key=lambda x:x[2])
print(sorted_pairs)
```
