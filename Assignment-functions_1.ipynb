```python
def febonacci():
    n = 7
    n1 = 0
    n2 = 1
    i = 1
    print(n1)
    print(n2)
    while i <= n-2: 
        n3 = n1 + n2
        print(n3,end=" ")
        n1 = n2
        n2 = n3
        i += 1
febonacci()        
```

```python
def evenno(n):
    for i in range(1,n+1):
        if i % 2 == 0:
            print(i,end=" ")
evenno(50)            
```

```python
def printlist(n):
    for i in n:
        print(i,end=" ")
my_list = [11,56,98,-44,"sid",True]        
printlist(my_list)        
```

```python
def factorial(n):
    fact = 1
    for i in range(1,n+1):
        fact = fact * i
    return fact
factorial(5)    
```

```python
def vowels(s):
    c = 0
    for i in s:
        if i == 'A' or i == 'E' or i == 'I' or i == 'O' or i == 'U' or i == 'a' or i == 'e' or i == 'i' or i == 'u':
            print(i,end=" ")
            c = c + 1
    return c
vowels("sidhharthlakshpriti")        
```

```python
def primeornot(num):
    if num <= 1:
        print("Not a prime number")
    else:
        i = 2
        is_prime = True
    while i < num:
        if num % i == 0:  
            is_prime = False
            break
        i += 1
    if is_prime:
        return True
    else:
        return False
primeornot(7)        
```
