```python
import numpy as np
def add(a, b):
    return a + b        
print(add(5, 7))
```

```python
a1 = np.array([10, 20, 30, 40])
a2 = np.array([2, 4, 5, 8])
print("Addition:", a1 + a2)
print("Subtraction:", a1 - a2)
print("Multiplication:", a1 * a2)
print("Division:", a1 / a2)
```

```python
a = np.array([3.14159, 2.71828, 1.23456, 9.87654])
r_a = np.round(a, 2)
print("Original Array:", a)
print("Rounded Array:", r_a)
```

```python
b = np.array([1, 2, 4, 10])
print("Natural Logarithm:", np.log(b))
```

```python
b = np.array([10, 20, 30, 40])
print("Sum:", np.sum(b))
```

```python
c = np.array([2, 3, 4, 5])
print("Product:", np.prod(c))
```

```python
d = np.array([10, 15, 25, 40])
print("Difference:", np.diff(d))
```

```python
a = 12
b = 18
print("LCM:", np.lcm(a, b))
```

```python
a = 12
b = 18
print("GCD:", np.gcd(a, b))
```

```python
ar = np.array([0, np.pi/2, np.pi])
print("Sin:", np.sin(ar))
print("Cos:", np.cos(ar))
print("Tan:", np.tan(ar))
```

```python
# 1.
# A ufunc (Universal Function) is a built-in NumPy function that performs element-wise operations on arrays efficiently. 
# It is faster than normal Python loops.

# 2.
# Ufuncs work directly on NumPy arrays.
# They are optimized for speed.
# They perform element-wise operations automatically.
# Regular Python functions usually work on one value at a time.

# 3.
# np.add()
# np.subtract()
# np.multiply()
# np.divide()
# np.power()
# np.mod()

# 4.
# Using:
# np.round()
# np.floor()
# np.ceil()
# np.trunc()
# Example:
# np.round(arr, 2)

# 5.
# np.log() → Natural logarithm (base e)
# np.log10() → Base 10 logarithm
# np.log2() → Base 2 logarithm

# 6.
# np.sum() returns the sum of all elements.
# np.prod() returns the product of all elements.
# Example:
# np.sum([1,2,3])   # 6
# np.prod([1,2,3])  # 6

# 7.
# Using np.diff().
# Example:
# np.diff([10,15,20])
# Output:
# [5 5]

# 8.
# np.gcd(a,b) returns the Greatest Common Divisor.
# np.lcm(a,b) returns the Least Common Multiple.
# Example:
# np.gcd(12,18)   # 6
# np.lcm(12,18)   # 36

# 9.
# np.sin()
# np.cos()
# np.tan()
# np.arcsin()
# np.arccos()
# np.arctan()

# 10.
# Hyperbolic ufuncs perform hyperbolic mathematical operations.
# Examples:
# np.sinh()
# np.cosh()
# np.tanh()
# They are commonly used in mathematics, physics, engineering, and machine learning.
```
