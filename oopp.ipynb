```python
from abc import ABC, abstractmethod

class employee(ABC):
    def __init__(self,name,salary):
        self.name = name
        self.__salary = salary
    def get_salary(self):
        return self.__salary
    @abstractmethod
    def cal_bonus(self):
        pass
        
class manager(employee):
    def cal_bonus(self):
        print(f'{self.name}')
        return self.get_salary() * 0.2
        
class developer(employee):
    def __init__(self,name,salary,projects):
        super().__init__(name,salary)
        self.projects = projects
    def cal_bonus(self):
        return self.get_salary() * 0.1 * self.projects

m = manager('Sakshi',50000)
print(m.cal_bonus())
d = developer('B',7000,12)
print(d.cal_bonus())

employee = [manager('Sid',100000), developer('Ankita',10000,10)]
for e in employee:
    print(e.cal_bonus())     
```

```python
from abc import ABC, abstractmethod
class Employee(ABC):
    def __init__(self,name):
        self.name = name
    @abstractmethod
    def cal_salary(self):
        pass
class FullTime(Employee):
    def __init__(self, name, salary):
        super().__init__(name)
        self.__salary = salary
    def cal_salary(self):
        return self.__salary
class PartTime(Employee):
    def __init__(self,name,hours,rate):
        super().__init__(name)
        self.hours = hours
        self.rate = rate
    def cal_salary(self):
        return self.hours * self.rate
p = PartTime('Aaksh',15,500)
print(p.cal_salary())
f = FullTime('Manoj',16)
print(f.cal_salary())
```

```python
from abc import ABC, abstractmethod
class SystemEntity(ABC):
    @abstractmethod
    def process(self):
        pass
class Account(SystemEntity):
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance
    def deposit(self, amt):
        self.__balance += amt
    def get_balance(self):
        return self.__balance
    def process(self):
        return f"{self.name} Balance: {self.__balance}"
class Product(SystemEntity):
    def __init__(self, name, price):
        self.name = name
        self.__price = price
    def get_price(self):
        return self.__price
    def process(self):
        return f"{self.name} Price: {self.__price}"
class Employee(SystemEntity):
    def __init__(self, name):
        self.name = name
    @abstractmethod
    def calculate(self):
        pass
    def process(self):
        return f"{self.name} Salary: {self.calculate()}"
class FullTime(Employee):
    def __init__(self, name, salary):
        super().__init__(name)
        self.__salary = salary
    def calculate(self):
        return self.__salary
class Vehicle(SystemEntity):
    def __init__(self, model, price):
        self.model = model
        self.__price = price
    def process(self):
        return f"{self.model} Rent: {self.__price}"
class LibraryItem(SystemEntity):
    def __init__(self, title):
        self.title = title
    def process(self):
        return f"Issued: {self.title}"
class Restaurant(SystemEntity):
    def __init__(self, name, bill):
        self.name = name
        self.bill = bill
    def process(self):
        return f"{self.name} Bill: {self.bill}"
class Payment(SystemEntity):
    def __init__(self, amount):
        self.amount = amount
    def process(self):
        return f"Paid {self.amount}"
class Patient(SystemEntity):
    def __init__(self, name, disease):
        self.name = name
        self.__disease = disease
    def process(self):
        return f"{self.name} has {self.__disease}"
class Course(SystemEntity):
    def __init__(self, title):
        self.title = title
    def process(self):
        return f"Enrolled in {self.title}"
class Device(SystemEntity):
    def __init__(self, name):
        self.name = name
        self.__state = False
    def toggle(self):
        self.__state = not self.__state
    def process(self):
        return f"{self.name} State Changed"
class MegaSystem:
    def __init__(self):
        self.modules = []

    def add_module(self, obj):
        self.modules.append(obj)

    def run(self):
        for m in self.modules:
            print(m.process())
system = MegaSystem()
system.add_module(Account("John", 10000))
system.add_module(Product("Laptop", 50000))
system.add_module(FullTime("Manoj", 30000))
system.add_module(Vehicle("Car", 2000))
system.add_module(LibraryItem("Python Book"))
system.add_module(Restaurant("Veg Hotel", 800))
system.add_module(Payment(500))
system.add_module(Patient("Amit", "Fever"))
system.add_module(Course("Data Science"))
system.add_module(Device("Light"))
system.run()
```

```python
from abc import ABC, abstractmethod
class Account(ABC):
    def __init__(self, name, accno, balance, pin):
        self.name = name
        self.accno = accno
        self.__balance = balance
        self.__pin = pin
        self._transactions = []
    def pin_authenticate(self, pin):
        return self.__pin == pin
    def deposit(self, amt):
        self.__balance += amt
        self._transactions.append(f"Deposited: {amt}")
    def get_balance(self):
        return self.__balance
    def show_transactions(self):
        return self._transactions
    @abstractmethod
    def withdraw(self, amt):
        pass
class SavingsAccount(Account):
    def withdraw(self, amt):
        if amt <= self.get_balance():
            self._Account__balance -= amt
            self._transactions.append(f"Withdrawn: {amt}")
        else:
            print("Insufficient Balance...")
class CurrentAccount(Account):
    def withdraw(self, amt):
        if amt <= self.get_balance() + 5000:
            self._Account__balance -= amt
            self._transactions.append(f"Withdrawn: {amt}")
        else:
            print("Overdraft Limit Exceeded")
acc = SavingsAccount("Sid", 101, 10000, 1234)
if acc.pin_authenticate(1234):
    acc.deposit(2000)
    acc.withdraw(5000)
    print("Balance:", acc.get_balance())
    print("Transactions:", acc.show_transactions())
else:
    print("Invalid PIN")
```

```python
from abc import ABC, abstractmethod
class User(ABC):
    def __init__(self, name):
        self.name = name
    @abstractmethod
    def role(self):
        pass
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
class Customer(User):
    def __init__(self, name, wallet):
        super().__init__(name)
        self.__wallet = wallet
        self.cart = []
    def add_to_cart(self, product):
        self.cart.append(product)
    def checkout(self):
        total = sum(p.price for p in self.cart)
        if total <= self.__wallet:
            self.__wallet -= total
            print("Order placed")
        else:
            print("Insufficient balance")
    def role(self):
        return "Customer"
class Seller(User):
    def __init__(self, name):
        super().__init__(name)
        self.__inventory = []
    def add_product(self, product):
        self.__inventory.append(product)
    def role(self):
        return "Seller"
class Admin(User):
    def role(self):
        return "Admin controlling system"
p1 = Product("Laptop", 50000)
cust = Customer("Sid", 60000)
cust.add_to_cart(p1)
cust.checkout()
```

```python
from abc import ABC, abstractmethod
class Ride(ABC):
    def __init__(self, distance):
        self.distance = distance
        self.__fare = 0
        self.__status = "Pending"
    @abstractmethod
    def calculate_fare(self):
        pass
    def complete_ride(self):
        self.__status = "Completed"
    def get_status(self):
        return self.__status
    def set_fare(self, fare):
        self.__fare = fare
    def get_fare(self):
        return self.__fare
class BikeRide(Ride):
    def calculate_fare(self):
        fare = self.distance * 10
        self.set_fare(fare)
        return fare
class CarRide(Ride):
    def calculate_fare(self):
        fare = self.distance * 20
        self.set_fare(fare)
        return fare
class LuxuryRide(Ride):
    def calculate_fare(self):
        fare = self.distance * 50
        self.set_fare(fare)
        return fare
class Customer:
    def __init__(self, name):
        self.name = name

    def book_ride(self, ride: Ride):
        fare = ride.calculate_fare()
        print(f"{self.name} booked ride. Fare: {fare}")
ride = CarRide(15)
cust = Customer("Sid")
cust.book_ride(ride)
print(ride.get_status())
ride.complete_ride()
print(ride.get_status())
```
