```python
# 1. 
# 1D Array → Single row of elements.
# [1,2,3]
# 2D Array → Rows and columns (like a table).
# [[1,2],
#  [3,4]]
# 3D Array → Collection of multiple 2D arrays.
# [[[1,2],[3,4]],
#  [[5,6],[7,8]]]

# 2.
# Python list slicing creates a new list (copy).
# NumPy slicing creates a view by default, so changes affect the original array.

# 3. 
# copy()	view()
# Creates a new array	Shares original data
# Changes do not affect original	Changes affect original
# Uses extra memory	Saves memory

# 4. How can you reshape arrays without changing the data?
# Use the reshape() method.
# Example:
# arr = np.arange(12)
# arr.reshape(3,4)
# Only the shape changes, not the values.

# 5. What are the advantages of NumPy arrays over Python lists?
# Faster execution
# Uses less memory
# Supports mathematical operations directly
# Easy reshaping and slicing
# Useful for data science and machine learning

# 6.
# Use np.where()
# arr = np.array([5,10,15,20])
# np.where(arr>10)
# Output
# (array([2,3]),)

# 7. 
# vstack() → Joins arrays vertically (adds rows).
# hstack() → Joins arrays horizontally (adds columns).
# concatenate() → General function that joins arrays along a specified axis.

# 8. 
# Use:
# np.nditer() → Iterates through every element efficiently.
# np.ndenumerate() → Gives both index and value.

# 9.
# Use Boolean Indexing.
# arr = np.array([1,2,3,4,5,6])
# print(arr[arr>3])
# Output
# [4 5 6]

# 10.
# View
# Shares the same memory as the original array.
# Faster.
# Less memory usage.
# Changes affect the original array.
# Copy
# Creates a completely new array.
# Uses extra memory.
# Changes do not affect the original array.
# Quick Revision (Interview Cheat Sheet)
# Function	Purpose
# np.array()	Create array
# np.ones()	Array of ones
# np.zeros()	Array of zeros
# np.arange()	Sequence of numbers
# np.linspace()	Evenly spaced numbers
# reshape()	Change shape
# shape	Get array shape
# size	Total elements
# ndim	Number of dimensions
# copy()	Independent copy
# view()	Shared memory
# concatenate()	Join arrays
# vstack()	Join vertically
# hstack()	Join horizontally
# split()	Split array
# where()	Find indices
# sort()	Sort array
# max()	Maximum value
# min()	Minimum value
# mean()	Average
# nditer()	Efficient iteration
# ndenumerate()	Iterate with index
```
