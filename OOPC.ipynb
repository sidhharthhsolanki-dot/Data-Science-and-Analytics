```python
from abc import ABC, abstractmethod

class BankAccount(ABC):

    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, amt):
        self.__balance = amt

    @abstractmethod
    def deposit(self, amt):
        pass

    @abstractmethod
    def withdrawl(self, amt):
        pass

    @abstractmethod
    def display_balance(self):
        pass


class SavingAccount(BankAccount):

    def deposit(self, amt):
        new_balance = self.get_balance() + amt
        self.set_balance(new_balance)

        print(f'{amt} deposited successfully.')

    def withdrawl(self, amt):

        if self.get_balance() - amt < 1000:
            print("Minimum 1000 balance is Required...")

        else:
            new_balance = self.get_balance() - amt
            self.set_balance(new_balance)

            print(f'{amt} is withdrawn successfully')

    def display_balance(self):
        print(f'Saving Account Balance: ₹{self.get_balance()}')


class CurrentAccount(BankAccount):

    def deposit(self, amt):

        new_balance = self.get_balance() + amt
        self.set_balance(new_balance)

        print(f'{amt} deposited successfully.')

    def withdrawl(self, amt):

        if self.get_balance() - amt < -5000:
            print("Overdraft limit exceeded.")

        else:
            new_balance = self.get_balance() - amt
            self.set_balance(new_balance)

            print(f'{amt} is withdrawn successfully')

    def display_balance(self):
        print(f"Current Account Balance: ₹{self.get_balance()}")


# Objects
s1 = SavingAccount("Rahul", 5000)
c1 = CurrentAccount("Amit", 2000)

# Savings Account
print("----- Savings Account -----")

s1.deposit(2000)

s1.withdrawl(5500)

s1.display_balance()

# Current Account
print("\n----- Current Account -----")

c1.deposit(3000)

c1.withdrawl(9000)

c1.display_balance()
```

```python
from abc import ABC, abstractmethod
class LibraryItem(ABC):
    def __init__(self, title, item_id, available=True):
        self.title = title
        self.item_id = item_id
        # Encapsulation
        self.__available = available
    def get_availability(self):
        return self.__available
    def set_availability(self, status):
        self.__available = status
    # Abstract Methods
    @abstractmethod
    def borrow_item(self):
        pass
    @abstractmethod
    def return_item(self):
        pass
    @abstractmethod
    def display_info(self):
        pass
class Book(LibraryItem):
    def __init__(self, title, item_id, author):
        super().__init__(title, item_id)
        self.author = author
    # Polymorphism
    def borrow_item(self):
        if self.get_availability():
            self.set_availability(False)
            print(f'Book "{self.title}" borrowed for 14 days.')
        else:
            print(f'Book "{self.title}" is currently unavailable.')
    def return_item(self):
        self.set_availability(True)
        print(f'Book "{self.title}" returned successfully.')
    def display_info(self):
        print("\n----- Book Information -----")
        print(f"Title      : {self.title}")
        print(f"Item ID    : {self.item_id}")
        print(f"Author     : {self.author}")
        print(f"Available  : {self.get_availability()}")
class Magazine(LibraryItem):
    def __init__(self, title, item_id, issue_number):
        super().__init__(title, item_id)
        self.issue_number = issue_number
    # Polymorphism
    def borrow_item(self):
        if self.get_availability():
            self.set_availability(False)
            print(f'Magazine "{self.title}" borrowed for 3 days.')
        else:
            print(f'Magazine "{self.title}" is currently unavailable.')
    def return_item(self):
        self.set_availability(True)
        print(f'Magazine "{self.title}" returned successfully.')
    def display_info(self):
        print("\n----- Magazine Information -----")
        print(f"Title        : {self.title}")
        print(f"Item ID      : {self.item_id}")
        print(f"Issue Number : {self.issue_number}")
        print(f"Available    : {self.get_availability()}")
b1 = Book("Python Programming", 101, "Mark Lutz")
m1 = Magazine("Tech Monthly", 202, "Issue-45")
b1.display_info()
b1.borrow_item()
b1.display_info()
b1.return_item()
b1.display_info()
m1.display_info()
m1.borrow_item()
m1.display_info()
m1.return_item()
m1.display_info()
```

```python
from abc import ABC, abstractmethod


# Abstract Class
class Employee(ABC):
    def __init__(self, employee_id, name, salary):
        self.employee_id = employee_id
        self.name = name
        self.__salary = salary
    def get_salary(self):
        return self.__salary
    def set_salary(self, amount):
        self.__salary = amount
    @abstractmethod
    def calculate_salary(self):
        pass
    @abstractmethod
    def display_details(self):
        pass
class FullTimeEmployee(Employee):
    def __init__(self, employee_id, name, salary, bonus, department):
        super().__init__(employee_id, name, salary)
        self.bonus = bonus
        self.department = department
    # Polymorphism
    def calculate_salary(self):
        total_salary = self.get_salary() + self.bonus
        return total_salary
    def display_details(self):
        print("\n----- Full Time Employee -----")
        print(f"Employee ID : {self.employee_id}")
        print(f"Name        : {self.name}")
        print(f"Department  : {self.department}")
        print(f"Base Salary : ₹{self.get_salary()}")
        print(f"Bonus       : ₹{self.bonus}")
        print(f"Total Salary: ₹{self.calculate_salary()}")
class PartTimeEmployee(Employee):
    def __init__(self, employee_id, name,
                 hours_worked, hourly_rate):
        # Temporary salary = 0
        super().__init__(employee_id, name, 0)
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate
    # Polymorphism
    def calculate_salary(self):
        total_salary = self.hours_worked * self.hourly_rate
        return total_salary
    def display_details(self):
        print("\n----- Part Time Employee -----")
        print(f"Employee ID : {self.employee_id}")
        print(f"Name        : {self.name}")
        print(f"Hours Worked: {self.hours_worked}")
        print(f"Hourly Rate : {self.hourly_rate}")
        print(f"Total Salary: {self.calculate_salary()}")
e1 = FullTimeEmployee(101,"Rahul",50000,10000,"Data Analytics")
e2 = PartTimeEmployee(202,"Amit",120,500)
e1.display_details()
e2.display_details()
```
