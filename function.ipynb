```python
def no1to10(n):
    for i in range(1,n+1):
        print(f"{i} ",end = '')
no1to10(10)
```

```python
def even1to50(n):
    for i in range(0,n,2):
        print(f'{i} ',end=' ')
even1to50(50)
```

```python
def table(n):
    for i in range(1,11):
        print(f'{n} * {i} = {n * i}')
table(5)
```

```python
def no1to10list():
    n = int(input("Enter the Number: "))
    l = []
    for i in range(1,n+1):
        n1 = int(input())
        l.append(n1)
    return l
no1to10list()
```

```python
def sum1to100():
    sum1 = 0
    for i in range(1,101):
        sum1 += i
    return sum1
sum1to100()
```
