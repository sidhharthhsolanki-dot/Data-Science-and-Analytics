```python
# KNN Classification
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
# Training Data
X = np.array([
    [1, 2],
    [2, 3],
    [7, 8]
])
# Output Data
y = ["Red", "Red", "Blue"]
# Create the KNN Model
model = KNeighborsClassifier(n_neighbors=1)
# Train the model
model.fit(X, y)
# Test Point
test_point = np.array([[3, 4]])
# Prediction
prediction = model.predict(test_point)
# Display the result
print("Test Point:", test_point[0])
print("Predicted Class:", prediction[0])
```

```python
# Student Pass/Fail Prediction
# Test Point (5, 80)
# k = 3
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
# Training Data
X = np.array([[2, 60],[3, 65],[6, 85],[7, 90], [8, 92]])
# Output Data
y = np.array(["Fail", "Fail", "Pass", "Pass", "Pass"])
# Create the KNN Model
model = KNeighborsClassifier(n_neighbors=3)
# Train the model
model.fit(X, y)
# Test Point
test = np.array([[5, 80]])
# Prediction
prediction = model.predict(test)
# Display the result
print("Prediction:", prediction[0])
```
