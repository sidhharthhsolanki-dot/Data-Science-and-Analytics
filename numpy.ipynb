```python
import numpy as np
ori_array = np.array([10, 20, 30, 40, 50])
cop_array = ori_array.copy()
cop_array[0] = 999
print("Original Array:", ori_array)
print("Copied Array:  ", cop_array)
```

```python
ori_array = np.array([10, 20, 30, 40, 50])
cop_array = ori_array.view()
cop_array[0] = 999
print("Original Array:", ori_array)
print("View Array:  ", cop_array)
```

```python
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])
print(a + b)          # Output: [11 22 33 44] -> Or use np.add(a, b)
print(a - b)          # Output: [ 9 18 27 36] -> Or use np.subtract(a, b)
# Element-wise Multiplicatio
print(a * b)          # Output: [ 10  40  90 160] -> Or use np.multiply(a, b)
# Element-wise Division
print(a / b)          # Output: [10. 10. 10. 10.] -> Or use np.divide(a, b)

matrix1 = np.array([[1, 2], [3, 4]])
matrix2 = np.array([[5, 6], [7, 8]])
# Using the @ operator (recommended for Python 3)
print(matrix1 @ matrix2)  
# Output: [[19 22]
#          [43 50]]
# Alternative method
# print(np.dot(matrix1, matrix2))
```
