```python
lst = list(map(int, input("Enter numbers separated by space: ").split()))
first = second = float('-inf')
for num in lst:
    if num > first:
        second = first
        first = num
    elif num > second and num != first:
        second = num
if second == float('-inf'):
    print("No second largest element")
else:
    print("Second largest element is:", second)
```
