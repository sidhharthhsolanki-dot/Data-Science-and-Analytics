```python
# Q1. Explain the difference between if, if-else, and if-elif-else.
# if
# Used when you want to execute code only when a condition is true.
# age = 20
# if age >= 18:
#     print("Adult")
# if-else
# Provides two alternatives.
# age = 16
# if age >= 18:
#     print("Adult")
# else:
#     print("Minor")
# if-elif-else
# Used when there are multiple conditions.
# marks = 85
# if marks >= 90:
#     print("A")
# elif marks >= 75:
#     print("B")
# elif marks >= 60:
#     print("C")
# else:
#     print("D")
# Interview answer:
# if handles a single condition, if-else handles two possible paths, and if-elif-else is used when multiple mutually exclusive conditions need to be checked.

# Q2. What is nested if-else and when is it used?
# A nested if-else means putting one if statement inside another if or else block.
# Example:
# age = 25
# salary = 50000
# if age >= 18:
#     if salary >= 30000:
#         print("Eligible")
#     else:
#         print("Salary too low")
# else:
#     print("Age too low")
# Nested conditions are useful when the second condition should only be checked after the first condition is satisfied.

# Q3. Can multiple elif statements be used?
# Yes. Python allows multiple elif statements.
# Example:
# marks = 82
# if marks >= 90:
#     print("A+")
# elif marks >= 80:
#     print("A")
# elif marks >= 70:
#     print("B")
# elif marks >= 60:
#     print("C")
# elif marks >= 40:
#     print("D")
# else:
#     print("F")
# Python checks the conditions from top to bottom.

# Q4. How does Python evaluate conditions in if-elif-else?
# Python evaluates conditions from top to bottom.
# Example:
# marks = 85
# if marks >= 90:
#     print("A+")
# elif marks >= 75:
#     print("A")
# elif marks >= 60:
#     print("B")
# else:
#     print("C")
# Process:
# 85 >= 90 → False
# 85 >= 75 → True
# Print A
# Stop checking the remaining conditions.
# Important:
# Only the first true branch in an if-elif-else chain executes.

# Q5. What happens if none of the conditions is true?
# If there is an else, the else block executes.
# age = 15
# if age >= 18:
#     print("Adult")
# else:
#     print("Minor")
# Output:
# Minor
# If there is no else, Python simply skips the entire conditional block.
# age = 15
# if age >= 18:
#     print("Adult")
# print("Program continues")
# Output:
# Program continues

# Q6. How can you check multiple conditions in a single if statement?
# Use logical operators:
# and
# Both conditions must be true.
# age = 25
# if age >= 18 and age <= 60:
#     print("Eligible")
# or
# At least one condition must be true.
# num = 15
# if num % 3 == 0 or num % 5 == 0:
#     print("Valid")
# not
# Reverses a Boolean condition.
# is_student = False
# if not is_student:
#     print("Not a student")
    
# Q7. Explain short-circuit evaluation in conditional statements.
# Short-circuit evaluation means Python may stop evaluating a logical expression as soon as the final result is known.
# With and
# If the first condition is False, Python doesn't need to evaluate the second condition.
# x = 0
# if x != 0 and 10 / x > 2:
#     print("Yes")
# Because:
# x != 0
# is false, Python stops there. It doesn't evaluate 10 / x.
# With or
# If the first condition is True, Python doesn't need to evaluate the second condition.
# x = 10
# if x > 5 or 10 / 0 > 2:
#     print("Yes")
# Since x > 5 is true, Python stops evaluating the or expression.
# Interview answer:
# Short-circuit evaluation improves efficiency and can also prevent unnecessary or unsafe evaluations.

# Q8. Can an if statement exist without an else?
# Yes.
# An else is optional.
# age = 20
# if age >= 18:
#     print("Adult")
# If the condition is false, Python simply skips the block.

# Q9. How do you compare multiple variables efficiently using conditional statements?
# Python allows chained comparisons.
# Instead of:
# if age >= 18 and age <= 60:
#     print("Eligible")
# you can write:
# if 18 <= age <= 60:
#     print("Eligible")
# Similarly:
# a = 10
# b = 20
# c = 30
# if a < b < c:
#     print("Numbers are in ascending order")
# Output:
# Numbers are in ascending order
# This is a very useful Python feature and is commonly asked in interviews.

# Q10. Give a real-world scenario where nested if-else is necessary.
# Example: Loan eligibility
# Suppose a bank checks:
# Whether the applicant is old enough.
# If they are old enough, check their salary.
# If salary is sufficient, check their credit score.
# age = 30
# salary = 50000
# credit_score = 750
# if age >= 21:
#     if salary >= 30000:
#         if credit_score >= 700:
#             print("Loan approved")
#         else:
#             print("Credit score too low")
#     else:
#         print("Salary too low")
# else:
#     print("Age requirement not satisfied")
# Why nested if?

# The second check is meaningful only after the first requirement has been satisfied.

# Important Conditional Concepts for Data Analytics Interviews
# Concept	Example	Meaning
# if	if age >= 18:	Execute if condition is true
# else	else:	Alternative branch
# elif	elif marks >= 60:	Additional condition
# and	age >= 18 and salary > 30000	Both must be true
# or	age < 18 or age > 60	At least one must be true
# not	not is_valid	Reverses Boolean value
# ==	x == 10	Equal to
# !=	x != 10	Not equal
# >	x > 10	Greater than
# <	x < 10	Less than
# >=	x >= 10	Greater than/equal
# <=	x <= 10	Less than/equal
# in	x in [1,2,3]	Membership
# not in	x not in [1,2,3]	Not a member
# Nested if	if inside if	Dependent conditions
# Chained comparison	18 <= age <= 60	Multiple comparisons
```
