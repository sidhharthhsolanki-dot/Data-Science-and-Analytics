### 1. What are the different types of plots available in Matplotlib?

**Answer:**

Matplotlib supports many types of plots, including:

- Line plot — shows trends or changes over values.
- Scatter plot — shows the relationship between two variables.
- Bar chart — compares categories.
- Histogram — shows the frequency distribution of numerical data.
- Box plot — summarizes the distribution using quartiles, median, and possible outliers.
- Pie chart — shows proportions of a whole.
- Area plot — displays quantities over an interval.
- Stem plot — represents data using lines and markers.
- Step plot — displays values as steps.
- Error-bar plot — displays values together with uncertainty or error ranges.

### 2. How do you customize markers and line styles in a plot?

**Answer:**

Markers and line styles can be customized using `marker`, `linestyle`, and related plotting arguments.

Examples:

```python
plt.plot(x, y, marker='o', linestyle='--')
```

Here, `marker='o'` displays circular markers and `linestyle='--'` creates a dashed line.

A shorthand such as `plt.plot(x, y, '--o')` can also be used.

### 3. How can you add titles, axis labels, and grid to a plot?

**Answer:**

Use:

- `plt.title()` to add a title.
- `plt.xlabel()` to label the x-axis.
- `plt.ylabel()` to label the y-axis.
- `plt.grid(True)` to display the grid.

Example:

```python
plt.title('Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
```

### 4. Explain the difference between a scatter plot and a line plot.

**Answer:**

A **line plot** connects data points with lines and is commonly used to show trends or changes across an ordered x-axis.

A **scatter plot** displays individual points without connecting them by default. It is commonly used to study the relationship or distribution between two numerical variables.

Example:

```python
plt.plot(x, y)       # Line plot
plt.scatter(x, y)    # Scatter plot
```

### 5. How do you create and customize a box plot?

**Answer:**

A box plot can be created using `plt.boxplot()`.

Example:

```python
data = [10, 20, 15, 30, 25, 35, 40]
plt.boxplot(data)
plt.show()
```

A box plot summarizes the distribution using the median and quartiles and can also show potential outliers. It can be customized using arguments such as `vert`, `notch`, `showmeans`, and `patch_artist`.

### 6. How do you create subplots and control their layout?

**Answer:**

Use `plt.subplots()` to create multiple plots in one figure.

Example:

```python
fig, axes = plt.subplots(2, 2)

axes[0, 0].plot(x, y)
axes[0, 1].scatter(x, y)
axes[1, 0].bar(categories, values)
axes[1, 1].hist(data)

plt.tight_layout()
plt.show()
```

`plt.tight_layout()` helps prevent titles and labels from overlapping.

### 7. How can you customize bar charts (color, width, orientation)?

**Answer:**

A bar chart can be customized using arguments such as:

- `color` — changes bar color.
- `width` — controls the width of vertical bars.
- `height` — controls the height of horizontal bars.
- `align` — controls bar alignment.

Vertical bar example:

```python
plt.bar(products, sales, color='green', width=0.6)
```

For a horizontal bar chart:

```python
plt.barh(products, sales)
```

### 8. How do you create a histogram and choose the number of bins?

**Answer:**

A histogram is created using `plt.hist()`.

Example:

```python
plt.hist(data, bins=5)
plt.show()
```

The `bins` argument controls how many intervals are used to group the data. A small number of bins gives a more general distribution, while a larger number provides more detail. The appropriate choice depends on the dataset and the amount of detail needed.

### 9. How can you customize a pie chart with labels, percentages, and explode?

**Answer:**

Use `labels` for category names, `autopct` for percentages, and `explode` to separate one or more slices.

Example:

```python
labels = ['A', 'B', 'C', 'D']
sizes = [25, 35, 20, 20]
explode = (0, 0.1, 0, 0)

plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    explode=explode
)
plt.show()
```

Here, the second slice is exploded because it has the largest value.

### 10. How do you save a Matplotlib figure to a file?

**Answer:**

Use `plt.savefig()` before `plt.show()`.

Example:

```python
plt.plot(x, y)
plt.title('My Plot')
plt.savefig('my_plot.png')
plt.show()
```

The figure can be saved in formats such as PNG, JPG, SVG, or PDF depending on the filename extension.
