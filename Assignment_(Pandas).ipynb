```python
import numpy as np 
import pandas as pd
customer_orders = {
    "OrderID": [101, 102, 103, 104, 105],
    "CustomerName": ["John Doe", None, "Jane Smith", "Alice Johnson", "Bob Brown"],
    "OrderDate": ["2023-03-01", "2023-03-05", "2023-03-07", None, "2023-03-10"],
    "Product": ["Laptop", "Smartphone", None, "Tablet", "Laptop"],
    "Quantity": [1, 2, 1, 3, None],
    "Price": [1200, 800, 600, 300, 1200],
    "Discount": [None, 50, 0, 30, None],
    "Email": ["john@example.com", "mary[at]email.com", "jane@example.com", "alice@example.com", "bob@example"]
}
df = pd.DataFrame(customer_orders)
df['CustomerName'].fillna('Unknown',inplace=True)
df["OrderDate"] = pd.to_datetime(df["OrderDate"])
median_date = df["OrderDate"].median()
df["OrderDate"].fillna(median_date, inplace=True)
df['Product'].fillna(df['Product'].mode()[0],inplace=True)
df['Quantity'].fillna(1,inplace=True)
df["Discount"].fillna(0, inplace=True)
df["Email"] = df["Email"].str.replace(r"\[at\]", "@", regex=True)
df["Email"] = df["Email"].str.replace(r"(@example)$", "@",regex=True)
print(df)
```

```python
import pandas as pd
import numpy as np
employee_records = {
    "EmployeeID": [201, 202, 203, 204, 205],
    "Name": ["Alice", "Bob", "Charlie", None, "Eva"],
    "Department": ["HR", "Sales", None, "IT", "Finance"],
    "JoiningDate": ["2020-01-15", "2019-11-10", None, "2021-05-20", "2018-07-01"],
    "Salary": [50000, 60000, 55000, None, 62000],
    "Bonus": [5000, None, 3000, 2000, None],
    "Email": ["alice@company.com", "bob.company.com", "charlie@company.com",
              "david@company.com", "eva@company.com"]
}
df = pd.DataFrame(employee_records)
print("\nOriginal DataFrame:")
print(df)
df["Name"].fillna("Not Specified", inplace=True)
df["Department"].fillna("General", inplace=True)
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])
earliest_date = df["JoiningDate"].min()
df["JoiningDate"].fillna(earliest_date, inplace=True)
df["Salary"].fillna(df["Salary"].mean(), inplace=True)
df["Bonus"].fillna(0, inplace=True)
df["Email"] = df["Email"].apply(lambda x: x if "@" in x else x.replace(".", "@", 1))
df["TotalCompensation"] = df["Salary"] + df["Bonus"]
filtered = df[df["TotalCompensation"] > 60000]
print("\nCleaned DataFrame:")
print(df)
print("\nEmployees with TotalCompensation > 60000:")
print(filtered)
```

```python
import pandas as pd
import numpy as np
employee_records = {
    "EmployeeID": [201, 202, 203, 204, 205],
    "Name": ["Alice", "Bob", "Charlie", None, "Eva"],
    "Department": ["HR", "Sales", None, "IT", "Finance"],
    "JoiningDate": ["2020-01-15", "2019-11-10", None, "2021-05-20", "2018-07-01"],
    "Salary": [50000, 60000, 55000, None, 62000],
    "Bonus": [5000, None, 3000, 2000, None],
    "Email": ["alice@company.com", "bob.company.com", "charlie@company.com", "david@company.com", "eva@company.com"]
}
df = pd.DataFrame(employee_records)
df["Name"].fillna("Not Specified", inplace=True)
df["Department"].fillna("General", inplace=True)
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])
earliest_date = df["JoiningDate"].min()
df["JoiningDate"].fillna(earliest_date, inplace=True)
df["Salary"].fillna(df["Salary"].mean(), inplace=True)
df["Bonus"].fillna(0, inplace=True)
df["Email"] = df["Email"].str.replace(" ", "")
df["Email"] = df["Email"].str.replace("[at]", "@", regex=False)
df["Email"] = df["Email"].apply(lambda x: x if "@" in x else None)
df["TotalCompensation"] = df["Salary"] + df["Bonus"]
result = df[df["TotalCompensation"] > 60000]
print(result)
```
