```python
# Q31. What are mutable and immutable data types in Python? Give examples.
# Mutable data types can be changed after they are created.
# Examples:
# list
# set
# dictionary
# Example:
# numbers = [1, 2, 3]
# numbers[0] = 100
# print(numbers)
# Output:
# [100, 2, 3]
# Immutable data types cannot be changed after creation.
# Examples:
# int
# float
# bool
# str
# tuple
# Example:
# x = 10
# x = 20
# Here, Python doesn't modify the existing integer 10; instead, x is made to reference another integer object.
# Interview answer:
# Mutable objects can be modified after creation, whereas immutable objects cannot be modified after creation. Lists, sets, and dictionaries are mutable, while integers, strings, booleans, and tuples are immutable.

# Q32. What is the difference between a list and a tuple?
# List	Tuple
# Mutable	Immutable
# Uses []	Uses ()
# Elements can be changed	Elements cannot be changed
# Generally more flexible	Generally more memory-efficient
# More methods available	Fewer methods
# Example:
# my_list = [1, 2, 3]
# my_tuple = (1, 2, 3)

# Q33. How do sets differ from lists and tuples?
# A set is an unordered collection of unique elements.
# my_set = {1, 2, 3, 3}
# print(my_set)
# Result contains only:
# {1, 2, 3}
# Main differences:
# Set does not allow duplicate elements.
# Set is unordered.
# Set does not support indexing.
# Set is mutable.
# Sets are useful for mathematical operations such as union and intersection.
# Q34. Can dictionary keys be mutable? Explain.
# No. Dictionary keys must be hashable.
# Therefore, commonly:
# int
# float
# str
# tuple
# can be dictionary keys.
# But these cannot normally be dictionary keys:
# list
# set
# dictionary
# Example:
# student = {
#     "name": "Rahul",
#     101: "Python"
# }
# Valid.
# But:
# student = {
#     [1, 2]: "Python"
# }
# produces:
# TypeError: unhashable type: 'list'
# Interview answer:
# Dictionary keys must be immutable/hashable so that Python can reliably calculate their hash value. Mutable objects such as lists and dictionaries cannot be used as dictionary keys.

# Q35. What are Boolean values and how are they used in Python?
# Python has two Boolean values:
# True
# False
# They are used for logical conditions and decision-making.
# Example:
# age = 25
# print(age >= 18)
# Output:
# True
# Boolean operators include:
# and
# or
# not
# Example:
# age = 25


# print(age >= 18 and age <= 60)
# Q36. How do you convert one data type to another in Python?
# Python provides built-in functions for type conversion.
# Examples:
# int("10")
# float("10.5")
# str(100)
# list((1, 2, 3))
# tuple([1, 2, 3])
# set([1, 2, 2, 3])
# Example:
# age = "25"
# age = int(age)
# print(age)
# print(type(age))
# Output:
# 25
# <class 'int'>
# This is called type casting/type conversion.

# Q37. How do you find the length of a string, list, tuple, or dictionary?
# Use:
# len()
# Example:
# text = "Python"
# numbers = [1, 2, 3]
# tup = (10, 20, 30)
# student = {"name": "Rahul", "age": 22}
# print(len(text))
# print(len(numbers))
# print(len(tup))
# print(len(student))
# Output:
# 6
# 3
# 3
# 2
# For a dictionary, len() returns the number of key-value pairs.

# Q38. Explain the difference between is and ==.
# ==
# Checks whether two objects have equal values.
# a = [1, 2, 3]
# b = [1, 2, 3]
# print(a == b)
# Output:
# True
# is
# Checks whether two variables refer to the same object.
# print(a is b)
# Output:
# False
# because a and b are separate list objects.
# Interview answer:
# == checks value equality, while is checks object identity.


# Q39. How do you access elements in a nested dictionary or list?
# Nested list
# data = [
#     [10, 20], [30, 40]
# ]
# print(data[0][1])
# Output:

# 20
# Nested dictionary
# student = {
#     "name": "Rahul",
#     "marks": {
#         "Python": 90,
#         "SQL": 85
#     }
# }
# print(student["marks"]["Python"])
# Output:
# 90

# Q40. What is slicing in Python and which data types support it?
# Slicing is used to extract a portion of a sequence.
# Syntax:
# sequence[start:stop:step]
# Example:
# text = "Python"
# print(text[1:4])
# Output:
# yth
# Common data types that support slicing:
# String
# List
# Tuple
# Range
# Example:
# numbers = [10, 20, 30, 40, 50]
# print(numbers[1:4])
# Output:
# [20, 30, 40]
# Sets and dictionaries do not support normal positional slicing.
```
